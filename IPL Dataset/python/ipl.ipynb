```python
df_sorted = df.sort_values(["season", "date"])

max_streak = 0
current_streak = 0
previous_winner = None

for winner in df_sorted["winner"].dropna():

    if winner == previous_winner:
        current_streak += 1
    else:
        current_streak = 1

    max_streak = max(max_streak, current_streak)

    previous_winner = winner

print("Maximum Winning Streak:", max_streak)
```
